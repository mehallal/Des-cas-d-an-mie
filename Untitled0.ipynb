## Préparation des données pour le Machine Learning
Nous allons nettoyer les données, encoder les variables catégorielles, et définir nos variables explicatives ($X$) ainsi que la cible ($y$).

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# 1. Copie des données pour garder l'original intact
df_ml = df.copy()

# Suppression des espaces invisibles dans le nom des colonnes (ex: 'Gender ' -> 'Gender')
df_ml.columns = df_ml.columns.str.strip()

# 2. Gestion de la valeur manquante unique dans 'PCV/HCT' par la médiane
median_pcv = df_ml['PCV/HCT'].median()
df_ml['PCV/HCT'] = df_ml['PCV/HCT'].fillna(median_pcv)

# 3. Encodage de la variable qualitative 'Gender'
# On s'assure de convertir en chaîne de caractères d'abord pour éviter tout problème
df_ml['Gender'] = df_ml['Gender'].astype(str).str.strip()
label_encoder = LabelEncoder()
df_ml['Gender'] = label_encoder.fit_transform(df_ml['Gender'])

# 4. Définition de X et y
# y : Variable cible (Decision_Class)
y = df_ml['Decision_Class']

# X : Variables de prédiction (toutes les colonnes sauf la cible)
X = df_ml.drop(columns=['Decision_Class'])

print("Features (X) :")
print(X.columns.tolist())
print("\nTarget (y) : 'Decision_Class'")
```

## Division des données en Train / Test
Nous séparons le jeu de données : 80% pour l'entraînement du modèle et 20% pour l'évaluation de ses performances.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

print(f"Taille de l'ensemble d'entraînement (75%) : {X_train.shape[0]} lignes")
print(f"Taille de l'ensemble de test (25%) : {X_test.shape[0]} lignes")
```

## Entraînement et Évaluation du Modèle
Nous allons entraîner un modèle Random Forest et évaluer ses performances sur l'ensemble de test.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Ré-entraînement du modèle sur les nouvelles données (75% / 25%)
model = RandomForestClassifier(random_state=42, n_estimators=100)
model.fit(X_train, y_train)

# 2. Nouvelles prédictions cohérentes sur l'ensemble de test de 251 lignes
y_pred = model.predict(X_test)

# 3. Métrique d'évaluation principale
accuracy = accuracy_score(y_test, y_pred)
print(f"Nouvelle précision globale (Accuracy) : {accuracy:.2%}\n")

print("Nouveau rapport de classification -")
print(classification_report(y_test, y_pred))

# 4. Matrice de confusion mise à jour
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Normal (0)', 'Anémie (1)'], yticklabels=['Normal (0)', 'Anémie (1)'])
plt.xlabel('Prédiction')
plt.ylabel('Réalité')
plt.title('Matrice de Confusion (25% Test - 251 échantillons)')
plt.show()
```

```python
from sklearn.metrics import cohen_kappa_score, matthews_corrcoef, roc_auc_score

# Extraction des valeurs de la nouvelle matrice de confusion (taille 251)
tn, fp, fn, tp = cm.ravel()

# Calcul des métriques statistiques avancées adaptées aux normes médicales
sensibilite = tp / (tp + fn)
specificite = tn / (tn + fp)
mcc = matthews_corrcoef(y_test, y_pred)
kappa = cohen_kappa_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, model.predict_proba(X_test)[:, 1])

print("=== RAPPORT D'ÉVALUATION STATISTIQUE DU MODÈLE (75% Train / 25% Test) ===\n")
print(f"1. Précision Globale (Accuracy)              : {accuracy:.4f} (Soit {accuracy:.2%})")
print(f"2. Sensibilité (Recall / Vrais Positifs)      : {sensibilite:.4f} (Soit {sensibilite:.2%})")
print(f"3. Spécificité (TNR / Vrais Négatifs)        : {specificite:.4f} (Soit {specificite:.2%})")
print(f"4. Indice de Kappa de Cohen                  : {kappa:.4f}")
print(f"5. Coefficient de Corrélation de Matthews (MCC): {mcc:.4f}")
print(f"6. Aire sous la courbe ROC (AUC)             : {roc_auc:.4f}")
```

### Explication des Résultats selon les Normes Statistiques :

*   **Précision Globale (98.51%)** :
    Le modèle classe correctement 98,51 % des patients de test. C'est un excellent résultat de départ.

*   **Sensibilité / Rappel (97.67%)** :
    Dans le cadre du dépistage de l'anémie, c'est la métrique la plus importante. Elle indique que sur 100 patients réellement malades (anémiés), le modèle en détectera correctement près de 98. Seuls 2 % de faux négatifs (patients malades non détectés) sont à déplorer.

*   **Spécificité (99.13%)** :
    Elle indique que le modèle est capable de cibler les personnes saines avec une précision extrême. Sur 100 personnes saines, plus de 99 sont correctement identifiées comme telles (seulement 0,87 % de faux positifs).

*   **Kappa de Cohen (0.9691) & Matthews (0.9692)** :
    Selon les critères statistiques de Landis et Koch, un indice Kappa supérieur à **0.80** traduit une **concordance presque parfaite** (très largement supérieure au hasard). Le coefficient MCC proche de **0.97** (où 1.00 est une prédiction parfaite) confirme la robustesse et l'équilibre absolu du modèle sur les deux classes.
